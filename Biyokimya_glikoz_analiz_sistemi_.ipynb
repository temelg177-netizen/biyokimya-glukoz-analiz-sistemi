```python
# 🧪 Biyokimya Laboratuvarı - Akıllı Glukoz Analiz Sistemi
# Python ile laboratuvar sonuçlarının otomatik sınıflandırılması

hasta_sonuclari = [
    ["Hasta_001", 95],
    ["Hasta_002", 142],
    ["Hasta_003", 65],
    ["Hasta_004", 88],
    ["Hasta_005", 210]
]

kritik_yuksek = []
kritik_dusuk = []
normal = []

print("=== BİYOKİMYA LABORATUVARI ===")
print("GLUKOZ ANALİZ RAPORU\n")

for hasta in hasta_sonuclari:
    hasta_id = hasta[0]
    glukoz = hasta[1]

    if glukoz > 100:
        print(f"UYARI: {hasta_id} - {glukoz} mg/dL - YÜKSEK")
        kritik_yuksek.append(hasta_id)

    elif glukoz < 70:
        print(f"UYARI: {hasta_id} - {glukoz} mg/dL - DÜŞÜK")
        kritik_dusuk.append(hasta_id)

    else:
        print(f"NORMAL: {hasta_id} - {glukoz} mg/dL")
        normal.append(hasta_id)

print("\n=== ÖZET RAPOR ===")
print(f"Toplam hasta: {len(hasta_sonuclari)}")
print(f"Normal sonuç: {len(normal)}")
print(f"Yüksek sonuç: {len(kritik_yuksek)}")
print(f"Düşük sonuç: {len(kritik_dusuk)}")

print("\nAnaliz tamamlandı.")
```
